```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Scaling Code

x=np.array([[1,2],[3,4],[5,6]])
scaler=StandardScaler()
x_scaler=scaler.fit_transform(x)
print(x_scaler)
```

```python
plt.plot(x_scaler, marker=0)
plt.show()
```

```python
# Uncoding

data=['Red','Blue','Red','Blue']
le=LabelEncoder()
encoded=le.fit_transform(data)
print(encoded)
```

```python
plt.plot(encoded)
plt.show()
```

Probabilistic and ensemble learning

* Combine multiple models to create a accurate, robust and stable model than any single model alone

* Eg. of Probabilistic model-logistic regression(linear model:classification and regression ,For spam detection, success prediction, yes/no machine with linear brain.), naive bayes :assume all features are independent of each other given the class label(doctor diagnosing disease, multimedia project success prediction)    

**Ensemble Learning**

* Bagging : Bootstrap(create multiple random subset ot training Data) + arrgregating(combine predictions from models trained on each subset). Random-forest reducing

* Random Forest : Bagging + Random features selection.

**Bias-Variance and trade-off**

* used to remove overfitting/underfitting
* Bias-error due to overly simplistic models that don't capture data complexity(underfitting)

* Variance-error
  * Error due to overly complex models
  * Model learns noise and small fluctuations in training data
  * Performs well on training data but poorly on new data
  * Leads to overfitting
